Ejecutar desde la raíz del worktree. USD/gal US se divide entre 3.785411784; MPG × 1.609344 / 3.785411784 = km/L. No se combina MX con US ni se inventa FX.


```python
import json, math, hashlib
from pathlib import Path
root = Path.cwd()
data = json.loads((root / "docs/research/B2_FUEL_PRICE_PERFORMANCE_REFERENCE_2026-09-30.json").read_text(encoding="utf-8"))
gallon_liters = 3.785411784
mile_km = 1.609344
rows = []
for performance in data["performance"]:
    price = next(p for p in data["prices"] if p["country"] == performance["country"])
    per_liter = price["value"] if price["unit"] == "MXN/liter" else price["value"] / gallon_liters
    km_liter = performance["value"] if performance["unit"] == "km/liter" else performance["value"] * mile_km / gallon_liters
    rows.append({"id": performance["id"], "currency": "MXN" if performance["country"] == "MX" else "USD", "kmPerLiter": km_liter, "nativeCostPerKm": per_liter / km_liter})
    assert not performance["emptyConfirmed"]
    assert not performance["targetProfileMatched"]
assert math.isclose(rows[2]["nativeCostPerKm"], 5.955 / 9.5 / mile_km, rel_tol=1e-12)
assert math.isclose(rows[3]["nativeCostPerKm"], 5.955 / 7.9 / mile_km, rel_tol=1e-12)
assert math.isclose(505.9 / 53, 9.5, abs_tol=0.05)
assert math.isclose(1363.1 / 172.3, 7.9, abs_tol=0.05)
for pin in data["unchangedFiles"] + [s for s in data["sources"] if s["file"]]:
    assert hashlib.sha256((root / pin["file"]).read_bytes()).hexdigest() == pin["sha256"]
for card in data["countryCards"]:
    assert card["loadedKmPerLiter"] is None and card["emptyKmPerLiter"] is None
    assert card["pilotFuelTripUsd"] is None
assert data["adoptedValues"] == 0 and not data["calibrated"]
print(json.dumps({"status": "PASS", "purpose": "CROSS_PERIOD_PUBLIC_SENSITIVITY_NOT_BASE_INPUT", "rows": rows, "matchedFiles": 6, "pilotRates": None, "kernelExecution": False}, indent=2))
```
