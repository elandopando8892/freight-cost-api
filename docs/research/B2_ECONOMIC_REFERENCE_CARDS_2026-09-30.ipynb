```python
import json, hashlib
from pathlib import Path
root = Path.cwd()
manifest = json.loads((root / "docs/research/B2_ECONOMIC_REFERENCE_CARDS_2026-09-30.json").read_text(encoding="utf-8"))
expected = ["GENERAL_BASE","COST_CAPITAL","COST_INSURANCE","COST_PAYROLL","COST_COMPANY","UTILIZATION","LABOR","FUEL","COST_MAINT","COST_TIRES","FINANCE","RISK","TECHNICAL_MARGIN"]
assert len(manifest["cards"]) == 2
assert {c["country"] for c in manifest["cards"]} == {"MX", "US"}
for card in manifest["cards"]:
    assert [r["section"] for r in card["sectionRows"]] == expected
    assert all(r["economicAmount"] is None and r["subtotalEligible"] is False for r in card["sectionRows"])
    assert all(v is None for v in card["outputs"].values())
    assert all(r["assumptions"] and r["exclusions"] and r["missing"] for r in card["sectionRows"])
for source in manifest["sources"]:
    assert hashlib.sha256((root / source["file"]).read_bytes()).hexdigest() == source["sha256"]
    assert source["refreshedExternallyThisCut"] is False
assert manifest["sharedTrailer"]["rule"] == "ONE_PHYSICAL_CYCLE_THEN_ALLOCATE"
assert manifest["sharedTrailer"]["costBearer"] is None
assert manifest["trip"]["routeSelected"] is False
assert manifest["calibrated"] is False and manifest["quoteReady"] is False
assert manifest["adoptedValues"] == 0 and manifest["decision"]["businessGatesClosed"] == 0
assert all(v is False for v in manifest["externalChanges"].values())
result = {"cards": len(manifest["cards"]), "documentedSectionCountryRows": sum(len(c["sectionRows"]) for c in manifest["cards"]), "pinnedSourcesMatched": len(manifest["sources"]), "eligibleNumericSubtotals": 0, "adoptedValues": 0, "businessGatesClosed": 0}
print(json.dumps(result, ensure_ascii=False, indent=2))
```
